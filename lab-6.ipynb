```python
import pandas as pd 
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA
```

```python
data = {
    'Study_Hours' : [2,4,6,8,10],
    'Attendance' : [50,60,70,80,90],
    'Marks' : [40,50,60,70,80]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original Data")
df
```

```python
normalizer = MinMaxScaler()
normalized_data = normalizer.fit_transform(df)
```

```python
normalized_df = pd.DataFrame(normalized_data,columns = df.columns)
```

```python
print("Normalized Data")
normalized_df.round(2)
```

```python
standardrizer = StandardScaler()
standardrized_data = standardrizer.fit_transform(df)
```

```python
standardrized_df = pd.DataFrame(standardrized_data,columns = df.columns)
```

```python
print("Standardrized Data")
standardrized_df.round(2)
```

```python
pca = PCA(n_components=2)
```

```python
pca_data = pca.fit_transform(standardrized_data)
```

```python
pca_df = pd.DataFrame(pca_data,columns = ["PC1","PC2"])
```

```python
print("Data after PCA")
pca_df.round(2)
```
