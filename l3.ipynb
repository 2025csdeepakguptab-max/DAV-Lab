```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "student" :["s1","s2","s3","s4","s5","s6","s7","s8","s9","s10"],
    "studyhour" : [2,3,4,5,6,7,8,9,4,5],
    "attendence": [65,70,72,75,78,80,82,85,68,74],
    "assignmentm":[12,14,15,17,18,19,20,22,14,16],
    "internalmarks":[18,20,22,24,25,27,28,30,21,23],
    "final":[45,50,55,60,65,70,75,80,52,58]
}
df = pd.DataFrame(data)
print(df)
```

```python
plt.hist(df["final"], bins=5)
plt.xlabel("final marks")
plt.ylabel("number of student ")
plt.title("distribution of final marks")
plt.show()
```

```python
plt.boxplot(df["final"])
plt.xlabel("final marks")
plt.ylabel("marks")
plt.title("boxplot of final marks")
plt.show()
```

```python
plt.boxplot([
    df["studyhour"],
    df["attendence"],
    df["assignmentm"],
    df["internalmarks"],
    df["final"]
])
plt.xticks([1,2,3,4,5],["studyhour","attendence","assignmentm","internalmarks","final"])
plt.xlabel("subject")
plt.ylabel("values")
plt.title("boxplot of all subject")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.pairplot(
    df[[
        "studyhour",
        "attendence",
        "assignmentm",
        "internalmarks",
        "final"
    ]
    ]
)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.corr(numeric_only=True),annot=True)
plt.title("correlation heatmap")
plt.show()
```
