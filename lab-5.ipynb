```python
Q1 = df["Marks"].quantile(.25)
Q2 = df["Marks"].quantile(.50)
Q3 = df["Marks"].quantile(.75)
```

```python
IQR = Q3- Q1
```

```python
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR
```

```python
import pandas as pd

marks = [10, 12, 14, 16, 18, 20, 22, 24, 26, 100]

df = pd.DataFrame({"Marks": marks})

mean = df["Marks"].mean()
sd = df["Marks"].std(ddof=0)

df["Z_Score"] = (df["Marks"] - mean) / sd

threshold = 2.0
outliers = df[df["Z_Score"].abs() > threshold]

print("Dataset with Z-Scores:")
print(df)

print("\nOutliers found:")
print(outliers)
```
